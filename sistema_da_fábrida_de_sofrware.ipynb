```python

ETAPAS = ["Solicitação", "Análise", "Projeto", "Desenvolvimento", "Testes", "Homologação", "Entrega"]


projetos = [
    {
        "id": 1,
        "nome": "Sistema de Biblioteca",
        "responsavel": "João",
        "entrada": "Requisitos do cliente",
        "saida": "Sistema funcional",
        "etapa_idx": 3,
        "status": "Em andamento"
    }
]

def cadastrar_projeto():
    print("\n--- CADASTRO DE NOVO PROJETO ---")
    nome = input("Nome do projeto: ")
    responsavel = input("Responsável: ")
    entrada = input("Entrada (ex: Requisitos do cliente): ")
    saida = input("Saída (ex: Sistema funcional): ")

    novo_p = {
        "id": len(projetos) + 1,
        "nome": nome,
        "responsavel": responsavel,
        "entrada": entrada,
        "saida": saida,
        "etapa_idx": 0,
        "status": "Pendente"
    }
    projetos.append(novo_p)
    print(f"\n Projeto '{nome}' cadastrado na fábrica com sucesso!")

def listar_projetos():
    print("\n" + "="*60)
    print("  PROJETOS NA FÁBRICA DE SOFTWARE")
    print("="*60)
    if not projetos:
        print("Nenhum projeto cadastrado.")
        return

    for p in projetos:
        etapa_atual = ETAPAS[p["etapa_idx"]]
        print(f"\n[ID {p['id']}] Projeto: {p['nome']}")
        print(f"  Responsável : {p['responsavel']}")
        print(f"  Entrada     : {p['entrada']}")
        print(f"  Etapa Atual : {etapa_atual}")
        print(f"  Saída       : {p['saida']}")
        print(f"  Status      : {p['status']}")
        print("-" * 40)

def avancar_etapa():
    listar_projetos()
    if not projetos:
        return

    p_id = int(input("\nDigite o ID do projeto que deseja avançar a etapa: "))
    projeto = next((p for p in projetos if p["id"] == p_id), None)

    if projeto:
        if projeto["etapa_idx"] < len(ETAPAS) - 1:
            projeto["etapa_idx"] += 1
            etapa_nova = ETAPAS[projeto["etapa_idx"]]
            print(f"\n O projeto '{projeto['nome']}' avançou para a etapa: [{etapa_nova}]!")


            if etapa_nova == "Entrega":
                projeto["status"] = "Concluído"
            elif etapa_nova == "Testes":
                projeto["status"] = "Em testes"
            else:
                projeto["status"] = "Em andamento"
        else:
            print("\n O projeto já está na última etapa ('Entrega').")
    else:
        print("\n Projeto não encontrado.")

def alterar_status():
    listar_projetos()
    if not projetos:
        return

    p_id = int(input("\nDigite o ID do projeto que deseja alterar o status: "))
    projeto = next((p for p in projetos if p["id"] == p_id), None)

    if projeto:
        print("\nEscolha o novo status:")
        print("1. Pendente")
        print("2. Em andamento")
        print("3. Em testes")
        print("4. Concluído")

        op = input("Opção (1-4): ")
        status_map = {"1": "Pendente", "2": "Em andamento", "3": "Em testes", "4": "Concluído"}

        if op in status_map:
            projeto["status"] = status_map[op]
            print(f"\n Status alterado para '{projeto['status']}' com sucesso!")
        else:
            print("\n Opção inválida.")
    else:
        print("\n Projeto não encontrado.")

def main():
    while True:
        print("\n" + "="*40)
        print("   FÁBRICA DE SOFTWARE - MENU")
        print("="*40)
        print("1. Cadastrar Projeto")
        print("2. Visualizar Projetos (Ver Etapas, Entradas, Saídas e Status)")
        print("3. Avançar Etapa do Projeto")
        print("4. Alterar Status do Projeto")
        print("5. Sair")

        opcao = input("\nEscolha uma opção (1-5): ")

        if opcao == "1":
            cadastrar_projeto()
        elif opcao == "2":
            listar_projetos()
        elif opcao == "3":
            avancar_etapa()
        elif opcao == "4":
            alterar_status()
        elif opcao == "5":
            print("\nSaindo do sistema... Até logo!")
            break
        else:
            print("\nOpção inválida, tente novamente.")

if __name__ == "__main__":
    main()
```
